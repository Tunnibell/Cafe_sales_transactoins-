# Data collection

-The dataset was provided in a ZIP file containing a CSV file named dirty_cafe_sales.csv.

-It was loaded into Python using pandas for data processing and analysis.

```python
#Upload the file for the project.

df = pd.read_csv("/content/clean_cafe_sales.csv")
```

# 1. Most Popular Items (Bar Chart)

-This shows which product sells the most.


```python
# Identifying which products customers purchase most frequently.

# The bar chart shows that Coffee is the most frequently purchased item in the cafe.

item_sales = df.groupby("item")["quantity"].sum()

item_sales.plot(kind="bar")

plt.title("Most Popular Cafe Items")
plt.xlabel("Item")
plt.ylabel("Total Quantity Sold")

plt.show()
```

# 2. Revenue by Item

-This Shows which item generates the most money.

```python
# Although coffie is more popular, but salad generates higher total revenue.

revenue = df.groupby("item")["total_spent"].sum()

revenue.plot(kind="bar")

plt.title("Revenue by Item")
plt.xlabel("Item")
plt.ylabel("Total Revenue")

plt.show()
```

# 3. Payment Method Distribution (Pie Chart)

-This Shows how customers pay.

```python
# The chart shows that Most customers prefer paying with digital wallet and credit cards rather than cash.


payment = df["payment_method"].value_counts()

payment.plot(kind="pie", autopct="%1.1f%%")

plt.title("Payment Method Distribution")

plt.ylabel("")

plt.show()
```

# 4. Sales Trend Over Time (Line Chart)

-We use line chart to represent how sales changes over time.

-while trying represent my line chart i got error saying ValueError: time data "ERROR" doesn't match format.

-This means that This means that in the transaction_date column, some rows contain the word "ERROR" instead of a real date. as a

-As a data analyst you have to find a solution to your error so i used errors="coerce" when converting the column.

```python
# It helps Valid dates - converted to datetime

# Invalid values like "ERROR" - replaced with NaT (missing date)

df["transaction_date"] = pd.to_datetime(df["transaction_date"], errors="coerce")
```

```python
# Then Check how many bad dates exist

df["transaction_date"].isna().sum()
```

```python
# Then Remove rows with invalid dates

df = df.dropna(subset=["transaction_date"])
```

```python
# Then i was able to get the sales trend over time


daily_sales = df.groupby("transaction_date")["total_spent"].sum()

daily_sales.plot()

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Total Sales")

plt.show()
```

# 5. Sales by Location (Bar Chart)

-This shows whether customers prefer takeaway or in-store purchases.

```python
# The chart shows that In-store purchases generate a little more revenue than takeaway orders.

location_sales = df.groupby("location")["total_spent"].sum()

location_sales.plot(kind="bar")

plt.title("Sales by Location")
plt.xlabel("Location")
plt.ylabel("Total Revenue")

plt.show()
```
